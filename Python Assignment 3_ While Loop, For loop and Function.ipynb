```python
# 2.Prompt the User:
# Guess the number:
guess = int(input("Guess the number (between 1 and 10): "))

# Maximum number of guesses allowed:
attempts = 3
print("I have chosen a number between 1 and 10. You have", attempts, "attempts to guess it!")
```

```python
# 3.Implement the guessing logic and 4.Control Statements:
import random

# Generate random number between 1 and 10:
secret_number = random.randint(1, 10)

# Maximum number of valid attempts:
attempts = 3

while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

    # Check for out of range input:
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    # Reduce attempts for a valid guess:
    attempts -= 1

    if guess < secret_number:
        print("Too low. Try again.")
    elif guess > secret_number:
        print("Too high. Try again.")
    else:
        print("Congratulations! You guessed the correct number.")
        break
else:
    print("Better luck next time!")
```

MULTIPLICATION TABLE GENERATOR:

```python
# Prompt user for input:
number = int(input("Enter the number for which you want the multiplication table: "))

# Generate and display the multiplication table:
for i in range(1, 11):
    result = number * i
    print(number, "x", i, "=", result)
```

BMI CALCULATOR:

```python
# Define a function:
def calculate_bmi(weight, height):
    bmi = weight / (height * height)
    return bmi

# Prompt user for input:
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# Calculate BMI:
bmi_value = calculate_bmi(weight, height)

# Display the result:
print("Your BMI is:", round(bmi_value, 2))
```
